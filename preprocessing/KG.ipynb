```python
def stable_softmax(x):
  z = x - np.max(x, axis=-1, keepdims=True)
  numerator = np.exp(z)
  denominator = np.sum(numerator, axis=-1, keepdims=True)
  return numerator/denominator
```

```python
def normalisation(x):
  sum =  np.sum(x, axis=-1, keepdims=True)
  return x/sum
```

```python
def genre_normalisation(dic:dict):
  dic = dict(dic)
  arr = np.array(list(dic.values()))
  res = normalisation(arr)
  final = {}
  i=0
  for key in dic:
    final[key] = res[i]
    i+=1
  return final
```

```python
# 1. Indexes first (idempotent, cheap)
def create_indexes(tx):
    tx.run("CREATE INDEX genre_name_idx IF NOT EXISTS FOR (g:Genre) ON (g.name)")
    tx.run("CREATE INDEX shelf_name_idx IF NOT EXISTS FOR (s:Shelf) ON (s.name)")

# 2. Node creation
def create_genre_nodes(tx, names):
    tx.run("""
        UNWIND $names AS n
        MERGE (:Genre {name: n})
    """, names=names)

def create_shelf_nodes(tx, names):
    tx.run("""
        UNWIND $names AS n
        MERGE (:Shelf {name: n})
    """, names=names)
```

```python
def create_book_nodes(tx, books):
    tx.run("""
        UNWIND $books AS b
        MERGE (book:Book {book_id: b.book_id})
        SET book.isbn = b.isbn,
            book.isbn13 = b.isbn13,
            book.asin = b.asin,
            book.kindle_asin = b.kindle_asin,
            book.text_reviews_count = b.text_reviews_count,
            book.ratings_count = b.ratings_count,
            book.average_rating = b.average_rating,
            book.country_code = b.country_code,
            book.language_code = b.language_code,
            book.is_ebook = b.is_ebook,
            book.description = b.description,
            book.format = b.format,
            book.link = b.link,
            book.publisher = b.publisher,
            book.num_pages = b.num_pages,
            book.publication_day = b.publication_day,
            book.publication_month = b.publication_month,
            book.publication_year = b.publication_year,
            book.edition_information = b.edition_information,
            book.work_id = b.work_id,
            book.title = b.title,
            book.title_without_series = b.title_without_series
    """, books=books)
```

```python
def create_genre_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (b:Book {book_id: e.book_id})
        MATCH (g:Genre {name: e.genre})
        MERGE (b)-[r:HAS_GENRE]->(g)
        SET r.weight = e.weight
    """, edges=edges)

def create_shelf_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (b:Book {book_id: e.book_id})
        MATCH (s:Shelf {name: e.shelf})
        MERGE (b)-[r:HAS_SHELF]->(s)
        SET r.weight = e.weight
    """, edges=edges)
```

```python
def parse_rating_dist(s):
    """'5:1|4:1|3:1|2:0|1:0|total:3' -> dict"""
    if not s:
        return None
    parts = dict(p.split(":") for p in s.split("|"))
    return {k: int(v) for k, v in parts.items()}

works_df["rating_dist_parsed"] = works_df["rating_dist"].apply(parse_rating_dist)

# Numeric coercion
for col in ["books_count", "reviews_count", "text_reviews_count",
            "ratings_count", "ratings_sum",
            "original_publication_year", "original_publication_month",
            "original_publication_day"]:
    works_df[col] = pd.to_numeric(works_df[col], errors="coerce")
```

```python
def expand_rating_dist(rd):
    if not rd:
        return {"rd_5": None, "rd_4": None, "rd_3": None, "rd_2": None, "rd_1": None, "rd_total": None}
    return {
        "rd_5": rd.get("5"), "rd_4": rd.get("4"), "rd_3": rd.get("3"),
        "rd_2": rd.get("2"), "rd_1": rd.get("1"), "rd_total": rd.get("total"),
    }

rd_expanded = pd.DataFrame(works_df["rating_dist_parsed"].apply(expand_rating_dist).tolist())
works_df = pd.concat([works_df, rd_expanded], axis=1)
```

```python
def create_work_index(tx):
    tx.run("CREATE INDEX work_id_idx IF NOT EXISTS FOR (w:Work) ON (w.work_id)")

def create_work_nodes(tx, works):
    tx.run("""
        UNWIND $works AS w
        MERGE (work:Work {work_id: w.work_id})
        SET work.original_title = w.original_title,
            work.best_book_id = w.best_book_id,
            work.original_publication_year = w.original_publication_year,
            work.original_publication_month = w.original_publication_month,
            work.original_publication_day = w.original_publication_day,
            work.original_language_id = w.original_language_id,
            work.default_description_language_code = w.default_description_language_code,
            work.default_chaptering_book_id = w.default_chaptering_book_id,
            work.media_type = w.media_type,
            work.books_count = w.books_count,
            work.reviews_count = w.reviews_count,
            work.text_reviews_count = w.text_reviews_count,
            work.ratings_count = w.ratings_count,
            work.ratings_sum = w.ratings_sum,
            work.rating_dist = w.rating_dist,
            work.rd_5 = w.rd_5,
            work.rd_4 = w.rd_4,
            work.rd_3 = w.rd_3,
            work.rd_2 = w.rd_2,
            work.rd_1 = w.rd_1,
            work.rd_total = w.rd_total
    """, works=works)

work_cols = [
    "work_id", "best_book_id", "original_title",
    "original_publication_year", "original_publication_month", "original_publication_day",
    "original_language_id", "default_description_language_code", "default_chaptering_book_id",
    "media_type", "books_count", "reviews_count", "text_reviews_count",
    "ratings_count", "ratings_sum", "rating_dist",
    "rd_5", "rd_4", "rd_3", "rd_2", "rd_1", "rd_total",
]

works_payload = (
    works_df[work_cols]
    .astype(object)
    .where(lambda x: x.notna(), None)
    .to_dict("records")
)

with driver.session(database="goodreads-poetry") as session:
    session.execute_write(create_work_index)
    for chunk in chunked(works_payload, 10_000):
        session.execute_write(create_work_nodes, chunk)

print(f"created {len(works_payload)} work nodes")
```

```python
def create_edition_of_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (b:Book {book_id: e.book_id})
        MATCH (w:Work {work_id: e.work_id})
        MERGE (b)-[:EDITION_OF]->(w)
    """, edges=edges)

edition_edges = (
    df[["book_id", "work_id"]]
    .dropna()
    .astype(str)
    .to_dict("records")
)

with driver.session(database="goodreads-poetry") as session:
    for chunk in chunked(edition_edges, 10_000):
        session.execute_write(create_edition_of_edges, chunk)

print(f"created {len(edition_edges)} edition_of edges")
```

```python
def create_authored_by_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (b:Book {book_id: e.book_id})
        MATCH (a:Author {author_id: e.author_id})
        MERGE (b)-[r:AUTHORED_BY]->(a)
        SET r.role = e.role,
            r.position = e.position
    """, edges=edges)

authored_edges = [
    {
        "book_id": str(row.book_id),
        "author_id": a["author_id"],
        "role": a.get("role", "") or None,  # store None for empty roles
        "position": i,                       # 0 = first author, 1 = second, etc.
    }
    for row in df.itertuples()
    for i, a in enumerate(row.authors or [])
]

print(f"prepared {len(authored_edges)} authored_by edges")

with driver.session(database=DB_NAME) as session:
    for chunk in chunked(authored_edges, 10_000):
        session.execute_write(create_authored_by_edges, chunk)
```

```python
def parse_gr_date(s):
    if not s or pd.isna(s):
        return None
    try:
        return pd.to_datetime(s, format="%a %b %d %H:%M:%S %z %Y")
    except (ValueError, TypeError):
        return None
```

```python
def create_wrote_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (u:User {user_id: e.user_id})
        MATCH (r:Review {review_id: e.review_id})
        MERGE (u)-[:WROTE]->(r)
    """, edges=edges)

def create_reviews_edges(tx, edges):
    tx.run("""
        UNWIND $edges AS e
        MATCH (r:Review {review_id: e.review_id})
        MATCH (b:Book {book_id: e.book_id})
        MERGE (r)-[:REVIEWS]->(b)
    """, edges=edges)

wrote_edges = reviews_with_text[["user_id", "review_id"]].to_dict("records")
reviews_edges = reviews_with_text[["review_id", "book_id"]].astype(str).to_dict("records")

with driver.session(database=DB_NAME) as session:
    for chunk in chunked(wrote_edges, 10_000):
        session.execute_write(create_wrote_edges, chunk)
    for chunk in chunked(reviews_edges, 10_000):
        session.execute_write(create_reviews_edges, chunk)
```

```python
# ----- 1. Pull interactions for our books -----------------------------------

poetry_db = client["poetry"]

book_ids = df["book_id"].astype(str).unique().tolist()
print(f"filtering interactions for {len(book_ids)} books")

BATCH = 5000
interactions_chunks = []
for i in range(0, len(book_ids), BATCH):
    chunk_ids = book_ids[i:i + BATCH]
    cursor = poetry_db.interactions.find(
        {"book_id": {"$in": chunk_ids}},
        {
            "_id": 0,
            "user_id": 1, "book_id": 1, "review_id": 1,
            "is_read": 1, "rating": 1, "date_added": 1,
        }
    )
    interactions_chunks.append(pd.DataFrame(list(cursor)))

interactions_df = pd.concat(interactions_chunks, ignore_index=True)
print(f"got {len(interactions_df)} interactions")
print(f"unique users: {interactions_df['user_id'].nunique()}")
```

```python
interactions_df["rating"] = pd.to_numeric(interactions_df["rating"], errors="coerce").fillna(0).astype(int)
interactions_df["is_read"] = interactions_df["is_read"].astype(bool)
interactions_df["book_id"] = interactions_df["book_id"].astype(str)
interactions_df["date_added"] = interactions_df["date_added"].apply(parse_gr_date)
interactions_df["date_added"] = pd.to_datetime(
    interactions_df["date_added"],
    format="%a %b %d %H:%M:%S %z %Y",
    errors="coerce",
    utc=True,  # normalize all timezones to UTC so the column has a uniform tz
)
interactions_df["date_added"] = interactions_df["date_added"].dt.strftime("%Y-%m-%dT%H:%M:%S%z")
interactions_df["date_added"] = interactions_df["date_added"].where(interactions_df["date_added"].notna(), None)

def assign_edge_type(row):
    if row["is_read"]:
        if row["rating"] >= 4:
            return "RATED_HIGH"
        if row["rating"] >= 1:
            return "RATED_LOW"
        return "READ_UNRATED"
    return "SHELVED"  # is_read=false, rating=0 by elimination

interactions_df["edge_type"] = interactions_df.apply(assign_edge_type, axis=1)

print("\nedge type distribution:")
print(interactions_df["edge_type"].value_counts())
```

```python
def create_user_book_edges(tx, edges, edge_type):
    # Edge type can't be a Cypher parameter — must be string-formatted.
    # Safe because edge_type is from a fixed allowlist below.
    tx.run(f"""
        UNWIND $edges AS e
        MATCH (u:User {{user_id: e.user_id}})
        MATCH (b:Book {{book_id: e.book_id}})
        MERGE (u)-[r:{edge_type}]->(b)
        SET r.rating = e.rating,
            r.date_added = e.date_added
    """, edges=edges)

ALLOWED_EDGE_TYPES = {"RATED_HIGH", "RATED_LOW", "READ_UNRATED", "SHELVED"}
edge_cols = ["user_id", "book_id", "rating", "date_added"]

for edge_type, group in interactions_df.groupby("edge_type"):
    assert edge_type in ALLOWED_EDGE_TYPES, f"unexpected edge type: {edge_type}"
    edges = (
        group[edge_cols]
        .astype(object)
        .where(lambda x: x.notna(), None)
        .to_dict("records")
    )
    print(f"creating {len(edges):>8d} {edge_type} edges...")
    with driver.session(database=DB_NAME) as session:
        for chunk in chunked(edges, 10_000):
            session.execute_write(create_user_book_edges, chunk, edge_type)

print("\ndone")
```
